```python
# =========================
# 4. MANUAL CONVOLUTION
# =========================

def manual_convolution_stride1(image, kernel):
    h, w = image.shape
    kh, kw = kernel.shape

    # Output size (no padding, stride = 1)
    output = np.zeros((h - kh + 1, w - kw + 1))

    # Slide filter
    for i in range(output.shape[0]):
        for j in range(output.shape[1]):

            # Extract 3x3 patch
            patch = image[i:i+kh, j:j+kw]

            # Element-wise multiply + sum
            output[i, j] = np.sum(patch * kernel)

    return output
```

```python
# =========================
# MANUAL CONVOLUTION (STRIDE = 2)
# =========================

def manual_convolution_stride2(image, kernel):
    h, w = image.shape
    kh, kw = kernel.shape

    # Output size for stride = 2
    out_h = ((h - kh) // 2) + 1
    out_w = ((w - kw) // 2) + 1

    output = np.zeros((out_h, out_w))

    i_out = 0
    for i in range(0, h - kh + 1, 2):   # stride = 2
        j_out = 0
        for j in range(0, w - kw + 1, 2):  # stride = 2

            patch = image[i:i+kh, j:j+kw]
            output[i_out, j_out] = np.sum(patch * kernel)

            j_out += 1

        i_out += 1

    return output
```

```python
# =========================
# MANUAL CONVOLUTION (STRIDE = 3)
# =========================

def manual_convolution_stride3(image, kernel):
    h, w = image.shape
    kh, kw = kernel.shape

    # Output size for stride = 3
    out_h = ((h - kh) // 3) + 1
    out_w = ((w - kw) // 3) + 1

    output = np.zeros((out_h, out_w))

    i_out = 0
    for i in range(0, h - kh + 1, 3):   # stride = 3
        j_out = 0
        for j in range(0, w - kw + 1, 3):  # stride = 3

            patch = image[i:i+kh, j:j+kw]
            output[i_out, j_out] = np.sum(patch * kernel)

            j_out += 1

        i_out += 1

    return output
```

```python
def conv2d(image, kernel, stride=1):
    h, w = image.shape
    kh, kw = kernel.shape

    out_h = ((h - kh) // stride) + 1
    out_w = ((w - kw) // stride) + 1

    output = np.zeros((out_h, out_w))

    i_out = 0
    for i in range(0, h - kh + 1, stride):
        j_out = 0
        for j in range(0, w - kw + 1, stride):
            patch = image[i:i+kh, j:j+kw]
            output[i_out, j_out] = np.sum(patch * kernel)
            j_out += 1
        i_out += 1

    return output
```

```python
def rgb_conv(R, G, B, KR, KG, KB, stride=1):
    out_R = conv2d(R, KR, stride)
    out_G = conv2d(G, KG, stride)
    out_B = conv2d(B, KB, stride)

    return out_R + out_G + out_B
```

```python
## Dimesions Calculations
import math
def conv_output_size(H, W, KH, KW, stride, padding=0):
    """
    Returns output height and width for convolution
    """
    out_h = (H - KH + 2 * padding) // stride + 1
    out_w = (W - KW + 2 * padding) // stride + 1
    return out_h, out_w
```

```python
def multi_stride_output(H, W, KH, KW, filters=1, padding=0):
    """
    Calculates output for stride = 1, 2, 3
    """

    results = {}

    for s in [1, 2, 3]:
        out_h, out_w = conv_output_size(H, W, KH, KW, s, padding)

        # final CNN output shape includes filters (depth)
        results[f"stride_{s}"] = (out_h, out_w, filters)

    return results
```

```python
import numpy as np
import cv2
```

```python
# =========================
# GRAYSCALE CONVOLUTION
# =========================
def conv2d_gray(image, kernel, stride=1):

    H, W = image.shape
    KH, KW = kernel.shape

    out_h = (H - KH) // stride + 1
    out_w = (W - KW) // stride + 1

    output = np.zeros((out_h, out_w))

    i_out = 0
    for i in range(0, H - KH + 1, stride):
        j_out = 0
        for j in range(0, W - KW + 1, stride):

            patch = image[i:i+KH, j:j+KW]
            output[i_out, j_out] = np.sum(patch * kernel)

            j_out += 1
        i_out += 1

    return output
```

```python
# =========================
# RGB CONVOLUTION
# =========================
def conv2d_rgb(image, kernel, stride=1):

    H, W, C = image.shape
    KH, KW, KC = kernel.shape

    out_h = (H - KH) // stride + 1
    out_w = (W - KW) // stride + 1

    output = np.zeros((out_h, out_w))

    i_out = 0
    for i in range(0, H - KH + 1, stride):
        j_out = 0
        for j in range(0, W - KW + 1, stride):

            patch = image[i:i+KH, j:j+KW, :]
            output[i_out, j_out] = np.sum(patch * kernel)

            j_out += 1
        i_out += 1

    return output
```

```python
# =========================
# RELU
# =========================
def relu(x):
    return np.maximum(0, x)
```

```python
# =========================
# POOLING
# =========================
def pooling(x, size=2, stride=2, pool_type="max"):

    H, W = x.shape

    out_h = (H - size) // stride + 1
    out_w = (W - size) // stride + 1

    output = np.zeros((out_h, out_w))

    i_out = 0
    for i in range(0, H - size + 1, stride):
        j_out = 0
        for j in range(0, W - size + 1, stride):

            patch = x[i:i+size, j:j+size]

            if pool_type == "max":
                output[i_out, j_out] = np.max(patch)
            else:
                output[i_out, j_out] = np.mean(patch)

            j_out += 1
        i_out += 1

    return output
```

```python
# =========================
# FLATTEN
# =========================
def flatten(x):
    return x.reshape(-1)
```

```python
# =========================
# SIGMOID
# =========================
def sigmoid(x):
    x = np.clip(x, -500, 500)
    return 1 / (1 + np.exp(-x))
```

```python
# =========================
# CNN PIPELINE
# =========================
def cnn(image, mode, num_filters, KH, stride):

    show(image, "Input Image")
    print("Input Shape:", image.shape)

    pooled_maps = []

    # =========================
    # FEATURE EXTRACTION
    # =========================
    for i in range(num_filters):

        kernel = np.random.randn(KH, KH) if mode == "gray" else np.random.randn(KH, KH, 3)

        # Convolution
        if mode == "gray":
            conv = conv2d_gray(image, kernel, stride)
        else:
            conv = conv2d_rgb(image, kernel, stride)

        show(conv, f"Conv Feature Map {i+1}")
        print("Conv Shape:", conv.shape)

        # ReLU
        relu_out = relu(conv)
        show(relu_out, f"ReLU Feature Map {i+1}")

        # Pooling
        pool_out = pooling(relu_out, 2, 2)
        show(pool_out, f"Pooling Feature Map {i+1}")
        print("Pooling Shape:", pool_out.shape)

        pooled_maps.append(pool_out)

    # =========================
    # CLASSIFICATION
    # =========================

    features = np.stack(pooled_maps, axis=-1)
    print("\nFinal Feature Tensor Shape:", features.shape)

    flat = flatten(features)
    print("Flatten Shape:", flat.shape)

    W = np.random.randn(flat.shape[0])
    b = np.random.randn(1)

    z = np.dot(flat, W) + b
    pred = sigmoid(z)

    print("\n================ PREDICTION ================")
    print("Probability:", pred)

    if pred > 0.5:
        print("Class: 1 (Object Present)")
    else:
        print("Class: 0 (No Object)")

    return pred
```

```python
# =========================
# INPUT SECTION
# =========================

import matplotlib.pyplot as plt   # for visualization

# Ask user for image path
img_path = input("Enter image path: ")

# Ask user for image type
mode = input("Gray or RGB? ").lower()

# Ask CNN parameters
num_filters = int(input("Number of filters: "))
KH = int(input("Kernel size: "))
stride = int(input("Stride: "))
```

```python
# =========================
# LOAD IMAGE
# =========================

if mode == "gray":
    image = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
else:
    image = cv2.imread(img_path)

# Check image
if image is None:
    print("Image not found!")
    exit()
```
