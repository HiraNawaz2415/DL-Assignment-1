```python
# =========================================
# IMPORT LIBRARIES
# =========================================
# Import OpenCV for reading image
import cv2
# Import NumPy for matrix operations
import numpy as np
# Import matplotlib for displaying images
import matplotlib.pyplot as plt
```

```python
# =========================================
# STEP 1: LOAD IMAGE
# =========================================

# Read image from folder
img = cv2.imread("/content/image.jpg")
# OpenCV reads image in BGR format
# Copy image
image_rgb = img.copy()
# Convert BGR to RGB
img_rgb = img[:, :, ::-1]
# Swap Blue and Red channels manually
#image_rgb[:, :, 0] = img[:, :, 2]   # Red channel
#image_rgb[:, :, 1] = img[:, :, 1]   # Green channel
#image_rgb[:, :, 2] = img[:, :, 0]   # Blue channel
```

```python
# =========================================
# STEP 2: SHOW RGB IMAGE
# =========================================
# Display RGB image
plt.imshow(image_rgb)
# Set title
plt.title("RGB Image")
# Hide axes
plt.axis("off")
# Show image
plt.show()
```

```python
# Print RGB matrix
print("RGB Matrix:\n")
print(img)
```

```python
# ====================================
# EXTRACT RGB CHANNELS
# ====================================
# Extract Red channel matrix
R = img_rgb[:, :, 0]
# Extract Green channel matrix
G = img_rgb[:, :, 1]
# Extract Blue channel matrix
B = img_rgb[:, :, 2]
```

```python
# Show Red channel
plt.imshow(R, cmap='Reds')
plt.title("Red Channel")
plt.axis("off")
plt.show()
```

```python
print("Red Matrix:\n")
print(R)
```

```python
# Show Green channel
plt.imshow(G, cmap='Greens')
plt.title("Green Channel")
plt.axis("off")
plt.show()
```

```python
print("\nGreen Matrix:\n")
print(G)
```

```python
# Show Blue channel
plt.imshow(B, cmap='Blues')
plt.title("Blue Channel")
plt.axis("off")
plt.show()
```

```python
print("\nBlue Matrix:\n")
print(B)
```

```python
# =========================================
# STEP 3: RGB IMAGE INFORMATION
# =========================================
# Get image dimensions
rows, cols, channels = image_rgb.shape
# Print dimensions
print("RGB Shape =", image_rgb.shape)
# RGB image storage calculation
# Each pixel contains 3 bytes
rgb_storage = rows * cols * 3
# Print storage
print("RGB Storage =", rgb_storage, "Bytes")
```

```python
# =========================================
# STEP 4: MANUAL RGB TO GRAYSCALE
# =========================================
# Create empty grayscale matrix
gray = np.zeros((rows, cols), dtype=np.uint8)
# Traverse each pixel
for i in range(rows):
    for j in range(cols):
        # Extract RGB values
        R = image_rgb[i, j, 0]
        G = image_rgb[i, j, 1]
        B = image_rgb[i, j, 2]
        # Apply grayscale formula manually
        gray_value = 0.299 * R + 0.587 * G + 0.114 * B
        # Store grayscale value
        gray[i, j] = gray_value
```

```python
# =========================================
# STEP 5: SHOW GRAYSCALE IMAGE
# =========================================
# Display grayscale image
plt.imshow(gray, cmap='gray')
# Set title
plt.title("Grayscale Image")
# Hide axes
plt.axis("off")
# Show image
plt.show()
```

```python
# Print grayscale matrix
print("Grayscale Matrix:\n")
print(gray)
```

```python
# =========================================
# STEP 6: GRAYSCALE INFORMATION
# =========================================
# Print grayscale dimensions
print("Grayscale Shape =", gray.shape)
# Storage calculation
# Grayscale uses 1 byte per pixel
gray_storage = rows * cols * 1
# Print grayscale storage
print("Grayscale Storage =", gray_storage, "Bytes")
```

```python
# =========================================
# STEP 7: MANUAL BLACK & WHITE CONVERSION
# =========================================
# Create empty binary matrix
binary = np.zeros((rows, cols), dtype=np.uint8)
# Traverse grayscale image
for i in range(rows):
    for j in range(cols):
        # Apply threshold manually
        if gray[i, j] > 127:
            # White pixel
            binary[i, j] = 255
        else:
            # Black pixel
            binary[i, j] = 0
```

```python
# =========================================
# STEP 8: SHOW BLACK & WHITE IMAGE
# =========================================
# Display binary image
plt.imshow(binary, cmap='gray')
# Set title
plt.title(" Black & White Image")
# Hide axes
plt.axis("off")
# Show image
plt.show()
```

```python
# Print black & white matrix
print("Black & White Matrix:\n")
print(binary)
```

```python
# =========================================
# STEP 9: BINARY IMAGE INFORMATION
# =========================================
# Print binary image dimensions
print("Binary Shape =", binary.shape)
# Binary image theoretically uses 1 bit per pixel
binary_storage_bits = rows * cols * 1
# Convert bits into bytes
binary_storage_bytes = binary_storage_bits / 8
# Print binary storage
print("Binary Storage =", binary_storage_bytes, "Bytes")
```

```python

# =========================================
# STEP 10: STORAGE COMPARISON
# =========================================
print("\n===== STORAGE COMPARISON =====")
# Labels
types = ["RGB", "Grayscale", "Binary"]
# Values in KB
values = [
    rgb_storage / 1024,
    gray_storage / 1024,
    binary_storage_bytes / 1024
]
# Make figure smaller (fix white space issue)
plt.figure(figsize=(6, 3))
# Horizontal bar chart
bars = plt.barh(types, values, color=["red", "gray", "black"])
# Title
plt.title("Storage Comparison of Image Types")
# X-axis label (storage values go horizontal)
plt.xlabel("Storage (KB)")
# Add value labels properly at end of bars
for bar in bars:
    plt.text(
        bar.get_width() + 0.5,   # move text slightly right
        bar.get_y() + bar.get_height()/2,
        f"{bar.get_width():.2f}",
        va='center'
    )
# Remove extra padding (IMPORTANT FIX)
plt.tight_layout()
# Show plot
plt.show()
# =========================================
# PRINT VALUES
# =========================================
print("RGB Storage =", rgb_storage / 1024, "KB")
print("Grayscale Storage =", gray_storage / 1024, "KB")
print("Binary Storage =", binary_storage_bytes / 1024, "KB")
```
