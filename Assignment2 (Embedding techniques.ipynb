# Embedding VS Encoding
Encoding assigns fixed numerical values
While embedding learns meaningful vector representations that capture similarity and context.
| Feature               | Encoding                                     | Embedding                                 |
| --------------------- | -------------------------------------------- | ----------------------------------------- |
| **Definition**        | Converts data into numbers using fixed rules | Converts data into vectors using learning |
| **Type**              | Rule-based                                   | Learning-based (neural networks)          |
| **Learning**          |  No training required                       |  Learned during training                 |
| **Meaning**           |  No semantic meaning                        |  Captures meaning & relationships        |
| **Vector Type**       | Sparse (many zeros)                          | Dense (compact values)                    |
| **Dimension**         | High (especially one-hot)                    | Low (compressed vectors)                  |
| **Similarity**        |  Cannot show similarity                     |  Similar items are closer                |
| **Context Awareness** |  No                                         |  Yes (in modern models)                  |
| **Flexibility**       | Fixed (does not change)                      | Dynamic (updates during training)         |
| **Examples**          | Label Encoding, One-Hot, Binary              | Word2Vec, GloVe, BERT                     |
| **Use Case**          | Simple ML models, preprocessing              | Deep learning, NLP, AI systems            |
| **Computation**       | Low                                          | Higher (needs training)                   |


##  Encoding and Embedding Techniques

#### Q1. Binary Encoding

```python
data = ['A', 'B', 'C', 'D']  # list of categories
for i, val in enumerate(data):  # assign index to each category
    binary = format(i, '02b')  # convert index to binary (2-bit)
    print(val, "→", binary)    # print category with binary value
```

#### Q2. Label Encoding

```python
data = ['Cat', 'Dog', 'Bird']  # input categories
unique = list(set(data))  # get unique values
encoding = {v: i for i, v in enumerate(unique)}  # assign number to each
print(encoding)  # print mapping (category → number)
```

#### Q3. One Hot Encoding

```python
data = ['Cat', 'Dog', 'Bird']  # categories
unique = list(set(data))  # unique categories
for val in data:
    # create vector: 1 for match, 0 otherwise
    vector = [1 if val == u else 0 for u in unique]
    print(val, "→", vector)  # print one-hot vector
```

#### Q4. TF

```python
sentence = "AI is AI"  # input text
words = sentence.split()  # split sentence into words
freq = {}  # empty dictionary for frequency
for w in words:
    freq[w] = freq.get(w, 0) + 1  # count occurrences of each word
print(freq)  # print word frequency
```

```python
from gensim.models import Word2Vec
# Import Word2Vec model from gensim library (used to create word embeddings)
# training sentences (input corpus)
sentences = [
    ["king", "queen", "man"],      # sentence 1 (words are tokenized)
    ["dog", "cat", "animal"],      # sentence 2
    ["king", "queen", "royal"]     # sentence 3
]
# train Word2Vec model → this is where learning happens
model = Word2Vec(
    sentences,       # input training data (sentences)
    vector_size=3,   # each word will be represented as 3-dimensional vector
    min_count=1      # include all words (even if they appear once)
)
# get learned embedding (vector representation of word "king")
print(model.wv["king"])
# model.wv → word vectors learned by model
# "king" → retrieves its learned numerical vector
```

```python
model.wv.most_similar("king")
```

#### 8.Elmo
embedding = f(word, sentence)

```python
# sentences with same word in different contexts
sentence1 = ["bank", "near", "river"]
sentence2 = ["bank", "gives", "loan"]

# simple function to simulate context-based meaning
def fake_elmo(sentence):
    embeddings = []

    for word in sentence:
        # simple rule-based "context understanding"
        if "river" in sentence and word == "bank":
            embeddings.append([0.9, 0.1])   # river bank meaning
        elif "loan" in sentence and word == "bank":
            embeddings.append([0.1, 0.9])   # financial bank meaning
        else:
            embeddings.append([0.5, 0.5])   # neutral

    return embeddings

# run for both sentences
print("Sentence 1 embeddings:", fake_elmo(sentence1))
print("Sentence 2 embeddings:", fake_elmo(sentence2))
```

#### 9. BERT

```python
sentence = "I love AI"  # input sentence

# Step 1: split sentence into tokens (basic preprocessing)
tokens = sentence.split()  # converts sentence → words

# Step 2: show tokens
print("Tokens:", tokens)

# Step 3: simulate BERT behavior (context understanding)
for i, word in enumerate(tokens):
    print("\nWord:", word)

    # BERT idea: each word is understood using full sentence context
    context_left = tokens[:i]     # words before
    context_right = tokens[i+1:]  # words after

    print("Left context:", context_left)
    print("Right context:", context_right)
    print("BERT uses FULL sentence → meaning is contextual")
```

#### 10. Sentence BERT

```python
# input sentences
s1 = "I love AI"
# first sentence (input text)

s2 = "AI is amazing"
# second sentence (input text)

# split into words (tokenization)
tokens1 = s1.split()
# converts sentence 1 into list of words

tokens2 = s2.split()
# converts sentence 2 into list of words

# print tokens
print("Sentence 1 tokens:", tokens1)
# shows words of sentence 1

print("Sentence 2 tokens:", tokens2)
# shows words of sentence 2

# simple idea of similarity (rule-based, not ML model)
common_words = set(tokens1) & set(tokens2)
# finds common words between both sentences using set intersection

print("\nCommon words:", common_words)
# displays words that are present in both sentences

# show explanation
print("\nBoth sentences have similar meaning because:")
# explanation heading

print("- Both talk about AI ")
# reason 1: both sentences contain AI topic

print("- Both express positive idea ")
# reason 2: both have positive sentiment

print("- Even if words are different, meaning is similar")
# reason 3: semantic similarity idea
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Step 1: Load pretrained sentence embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Step 2: Input sentences
s1 = "I love AI"
s2 = "AI is amazing"

# Step 3: Convert sentences into embeddings (vectors)
emb1 = model.encode([s1])
emb2 = model.encode([s2])

# Step 4: Print embeddings
print("Embedding 1:", emb1)
print("Embedding 2:", emb2)

# Step 5: Compute similarity between sentences
similarity = cosine_similarity(emb1, emb2)

print("\nSimilarity Score:", similarity[0][0])
```
